```python
columns_to_drop = df.filter(regex='^effectStickers').columns
columns_to_drop_2 = df.filter(regex='^videoMeta/subtitleLinks').columns
columns_to_drop_3 = ['videoMeta/coverUrl', 'videoMeta/height', 'videoMeta/originalCoverUrl', 'videoMeta/width', 'createTimeISO', 'Unnamed: 0']
columns_to_drop_4 = df.filter(regex='^detailedMentions').columns
columns_to_drop_5 = df.filter(regex='^mentions').columns
columns_to_drop_6 = ['authorMeta/avatar', 'authorMeta/originalAvatarUrl', 'authorMeta/bioLink', 'authorMeta/friends', 'authorMeta/privateAccount', 'isPinned']
columns_to_drop_7 = df.filter(regex='^slideshowImageLinks').columns
df_cleaned = df.drop(columns=columns_to_drop)
df_cleaned = df_cleaned.drop(columns=columns_to_drop_2)
df_cleaned = df_cleaned.drop(columns=columns_to_drop_3)
df_cleaned = df_cleaned.drop(columns=columns_to_drop_4)
df_cleaned = df_cleaned.drop(columns=columns_to_drop_5)
df_cleaned = df_cleaned.drop(columns=columns_to_drop_6)
df_cleaned = df_cleaned.drop(columns=columns_to_drop_7)

hashtag_columns = [col for col in df_cleaned.columns if 'hashtags' in col]
df_cleaned['hashtags'] = df_cleaned[hashtag_columns].apply(lambda row: [x for x in row if pd.notna(x)], axis=1)
df_cleaned = df_cleaned.drop(columns=hashtag_columns)

df_cleaned.shape
```

```python
df_cleaned = df_cleaned[df_cleaned['error'].isna()]
df_cleaned = df_cleaned.drop('error', axis=1)
df_cleaned.shape
```

```python
df_cleaned = df_cleaned[df_cleaned['authorMeta/fans'] != 0]
df_cleaned['viewsPerFollowers'] = df_cleaned['playCount']/df_cleaned['authorMeta/fans']
df_cleaned['likesPerFollowers'] = df_cleaned['diggCount']/df_cleaned['authorMeta/fans']
df_cleaned['commentsPerFollowers'] = df_cleaned['commentCount']/df_cleaned['authorMeta/fans']
df_cleaned['sharesPerFollowers'] = df_cleaned['shareCount']/df_cleaned['authorMeta/fans']
df_cleaned['savesPerFollowers'] = df_cleaned['collectCount']/df_cleaned['authorMeta/fans']

df_cleaned[['viewsPerFollowers', 'likesPerFollowers', 'commentsPerFollowers', 'sharesPerFollowers', 'savesPerFollowers']]

viewsPerFollowers_mean = df_cleaned['viewsPerFollowers'].mean(axis=0)
likesPerFollowers_mean = df_cleaned['likesPerFollowers'].mean(axis=0)
commentsPerFollowers_mean = df_cleaned['commentsPerFollowers'].mean(axis=0)
sharesPerFollowers_mean = df_cleaned['sharesPerFollowers'].mean(axis=0)
savesPerFollowers_mean = df_cleaned['savesPerFollowers'].mean(axis=0)
followers_mean = df_cleaned['authorMeta/fans'].mean(axis=0)

print(f'viewsPerFollowers_mean: {viewsPerFollowers_mean}')
print(f'likesPerFollowers_mean: {likesPerFollowers_mean}')
print(f'commentsPerFollowers_mean: {commentsPerFollowers_mean}')
print(f'sharesPerFollowers_mean: {sharesPerFollowers_mean}')
print(f'savesPerFollowers_mean: {savesPerFollowers_mean}')
print(f'followers mean: {followers_mean}')
```
